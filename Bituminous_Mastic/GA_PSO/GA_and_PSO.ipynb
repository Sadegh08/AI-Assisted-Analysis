```python
# ============================================================
# Bituminous Mastic – AI-Assisted Analysis
# Independent GA and PSO estimation of Aging Index coefficients
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)


# ============================================================
# 1. FILE AND COLUMN SETTINGS
# ============================================================

FILE_PATH = "/content/Aging index.xlsx"
SHEET_NAME = "Sheet1"

COL_FILLER = "Filler type"
COL_FB = "F/B by weight"
COL_FREQUENCY = "Frequency (rad/s)"
COL_AI = "Aging Index"


# ============================================================
# 2. OPTIMIZATION SETTINGS
# ============================================================

N_RUNS = 20

# Genetic Algorithm
GA_POPULATION_SIZE = 120
GA_GENERATIONS = 600
GA_CROSSOVER_RATE = 0.85
GA_MUTATION_PROBABILITY = 0.20
GA_ELITE_SIZE = 5
GA_TOURNAMENT_SIZE = 3

# Particle Swarm Optimization
PSO_SWARM_SIZE = 120
PSO_ITERATIONS = 600
PSO_INERTIA_START = 0.90
PSO_INERTIA_END = 0.40
PSO_COGNITIVE = 1.70
PSO_SOCIAL = 1.70

# Coefficient bounds
COEFFICIENT_LOWER_BOUND = -100.0
COEFFICIENT_UPPER_BOUND = 100.0

BASE_RANDOM_SEED = 42


# ============================================================
# 3. LOAD AND CHECK DATA
# ============================================================

df = pd.read_excel(
    FILE_PATH,
    sheet_name=SHEET_NAME
)

required_columns = [
    COL_FILLER,
    COL_FB,
    COL_FREQUENCY,
    COL_AI
]

missing_columns = [
    column
    for column in required_columns
    if column not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing columns: {missing_columns}\n"
        f"Available columns: {df.columns.tolist()}"
    )

data = df[required_columns].copy()

for column in [
    COL_FB,
    COL_FREQUENCY,
    COL_AI
]:
    data[column] = pd.to_numeric(
        data[column],
        errors="coerce"
    )

data = data.dropna().reset_index(drop=True)

if len(data) == 0:
    raise ValueError(
        "No valid observations remained after data cleaning."
    )

if (data[COL_FREQUENCY] <= 0).any():
    raise ValueError(
        "All frequency values must be greater than zero "
        "before applying log10."
    )

print("Number of observations:", len(data))
print("Filler types:", data[COL_FILLER].unique())

display(data)


# ============================================================
# 4. PREPARE VARIABLES
# ============================================================

fb = data[
    COL_FB
].to_numpy(dtype=float)

frequency = data[
    COL_FREQUENCY
].to_numpy(dtype=float)

log_frequency = np.log10(
    frequency
)

y = data[
    COL_AI
].to_numpy(dtype=float)

filler_clean = (
    data[COL_FILLER]
    .astype(str)
    .str.strip()
    .str.lower()
)


# Rheofiller is the reference category.
#
# Rheofiller:
# D_cement = 0
# D_chromium = 0
#
# Cement:
# D_cement = 1
# D_chromium = 0
#
# Chromium steel:
# D_cement = 0
# D_chromium = 1

D_cement = (
    filler_clean.eq("cement")
    .astype(float)
    .to_numpy()
)

D_chromium = (
    filler_clean.eq("chromium steel")
    .astype(float)
    .to_numpy()
)

recognized_fillers = (
    filler_clean.eq("rheofiller")
    | filler_clean.eq("cement")
    | filler_clean.eq("chromium steel")
)

if not recognized_fillers.all():

    unknown_fillers = data.loc[
        ~recognized_fillers,
        COL_FILLER
    ].unique()

    raise ValueError(
        f"Unknown filler names detected: {unknown_fillers}"
    )

coding_check = pd.DataFrame({
    "Filler type": data[COL_FILLER],
    "D_cement": D_cement,
    "D_chromium": D_chromium
}).drop_duplicates()

print("\nDummy coding:")
display(coding_check)


# ============================================================
# 5. DESIGN MATRIX FOR THE EMPIRICAL EQUATION
# ============================================================

# AI_hat =
# beta_0
# + beta_1(f/b)
# + beta_2 log10(omega)
# + beta_3(f/b)^2
# + beta_4[log10(omega)]^2
# + beta_5(f/b)log10(omega)
# + beta_6 D_cement
# + beta_7 D_chromium

X = np.column_stack([
    np.ones(len(data)),
    fb,
    log_frequency,
    fb ** 2,
    log_frequency ** 2,
    fb * log_frequency,
    D_cement,
    D_chromium
])

coefficient_names = [
    "beta_0",
    "beta_1_f_b",
    "beta_2_log_frequency",
    "beta_3_f_b_squared",
    "beta_4_log_frequency_squared",
    "beta_5_interaction",
    "beta_6_cement",
    "beta_7_chromium_steel"
]

N_COEFFICIENTS = X.shape[1]

lower_bounds = np.full(
    N_COEFFICIENTS,
    COEFFICIENT_LOWER_BOUND,
    dtype=float
)

upper_bounds = np.full(
    N_COEFFICIENTS,
    COEFFICIENT_UPPER_BOUND,
    dtype=float
)


# ============================================================
# 6. OBJECTIVE FUNCTION AND METRICS
# ============================================================

def predict_ai(beta):

    beta = np.asarray(
        beta,
        dtype=float
    )

    if beta.shape != (N_COEFFICIENTS,):
        raise ValueError(
            f"Expected {N_COEFFICIENTS} coefficients, "
            f"received shape {beta.shape}."
        )

    return X @ beta


def rmse_objective(beta):

    prediction = predict_ai(
        beta
    )

    return float(
        np.sqrt(
            np.mean(
                (y - prediction) ** 2
            )
        )
    )


def calculate_metrics(beta):

    prediction = predict_ai(
        beta
    )

    rmse = np.sqrt(
        mean_squared_error(
            y,
            prediction
        )
    )

    mae = mean_absolute_error(
        y,
        prediction
    )

    r2 = r2_score(
        y,
        prediction
    )

    nonzero_mask = y != 0

    if nonzero_mask.any():

        mape = (
            np.mean(
                np.abs(
                    (
                        y[nonzero_mask]
                        - prediction[nonzero_mask]
                    )
                    / y[nonzero_mask]
                )
            )
            * 100
        )

    else:
        mape = np.nan

    return {
        "R2": float(r2),
        "RMSE": float(rmse),
        "MAE": float(mae),
        "MAPE (%)": float(mape)
    }


# ============================================================
# 7. GENETIC ALGORITHM
# ============================================================

def tournament_selection(
    population,
    fitness,
    rng
):

    candidate_indices = rng.choice(
        len(population),
        size=GA_TOURNAMENT_SIZE,
        replace=False
    )

    winner_index = candidate_indices[
        np.argmin(
            fitness[candidate_indices]
        )
    ]

    return population[
        winner_index
    ].copy()


def blend_crossover(
    parent_1,
    parent_2,
    rng
):

    alpha = rng.random(
        N_COEFFICIENTS
    )

    child_1 = (
        alpha * parent_1
        + (1.0 - alpha) * parent_2
    )

    child_2 = (
        alpha * parent_2
        + (1.0 - alpha) * parent_1
    )

    return child_1, child_2


def mutate(
    individual,
    generation,
    rng
):

    child = individual.copy()

    mutation_mask = (
        rng.random(N_COEFFICIENTS)
        < GA_MUTATION_PROBABILITY
    )

    progress = generation / max(
        GA_GENERATIONS - 1,
        1
    )

    initial_scale = 0.10
    final_scale = 0.002

    relative_scale = (
        initial_scale
        * (1.0 - progress)
        + final_scale
        * progress
    )

    mutation_std = (
        relative_scale
        * (
            upper_bounds
            - lower_bounds
        )
    )

    noise = rng.normal(
        loc=0.0,
        scale=mutation_std,
        size=N_COEFFICIENTS
    )

    child[mutation_mask] += noise[
        mutation_mask
    ]

    return np.clip(
        child,
        lower_bounds,
        upper_bounds
    )


def run_ga(seed):

    rng = np.random.default_rng(
        seed
    )

    population = rng.uniform(
        lower_bounds,
        upper_bounds,
        size=(
            GA_POPULATION_SIZE,
            N_COEFFICIENTS
        )
    )

    best_history = []

    for generation in range(
        GA_GENERATIONS
    ):

        fitness = np.array([
            rmse_objective(individual)
            for individual in population
        ])

        ranking = np.argsort(
            fitness
        )

        population = population[
            ranking
        ]

        fitness = fitness[
            ranking
        ]

        best_history.append(
            fitness[0]
        )

        new_population = [
            individual.copy()
            for individual
            in population[:GA_ELITE_SIZE]
        ]

        while (
            len(new_population)
            < GA_POPULATION_SIZE
        ):

            parent_1 = tournament_selection(
                population,
                fitness,
                rng
            )

            parent_2 = tournament_selection(
                population,
                fitness,
                rng
            )

            if (
                rng.random()
                < GA_CROSSOVER_RATE
            ):

                child_1, child_2 = blend_crossover(
                    parent_1,
                    parent_2,
                    rng
                )

            else:

                child_1 = parent_1.copy()
                child_2 = parent_2.copy()

            child_1 = mutate(
                child_1,
                generation,
                rng
            )

            child_2 = mutate(
                child_2,
                generation,
                rng
            )

            new_population.append(
                child_1
            )

            if (
                len(new_population)
                < GA_POPULATION_SIZE
            ):
                new_population.append(
                    child_2
                )

        population = np.asarray(
            new_population,
            dtype=float
        )

    final_fitness = np.array([
        rmse_objective(individual)
        for individual in population
    ])

    best_index = np.argmin(
        final_fitness
    )

    return (
        population[best_index].copy(),
        float(final_fitness[best_index]),
        np.asarray(best_history)
    )


# ============================================================
# 8. PARTICLE SWARM OPTIMIZATION
# ============================================================

def run_pso(seed):

    rng = np.random.default_rng(
        seed
    )

    positions = rng.uniform(
        lower_bounds,
        upper_bounds,
        size=(
            PSO_SWARM_SIZE,
            N_COEFFICIENTS
        )
    )

    search_range = (
        upper_bounds
        - lower_bounds
    )

    velocity_limit = (
        0.15 * search_range
    )

    velocities = rng.uniform(
        -velocity_limit,
        velocity_limit,
        size=(
            PSO_SWARM_SIZE,
            N_COEFFICIENTS
        )
    )

    personal_best_positions = (
        positions.copy()
    )

    personal_best_scores = np.array([
        rmse_objective(position)
        for position in positions
    ])

    global_best_index = np.argmin(
        personal_best_scores
    )

    global_best_position = (
        personal_best_positions[
            global_best_index
        ].copy()
    )

    global_best_score = float(
        personal_best_scores[
            global_best_index
        ]
    )

    best_history = []

    for iteration in range(
        PSO_ITERATIONS
    ):

        progress = iteration / max(
            PSO_ITERATIONS - 1,
            1
        )

        inertia = (
            PSO_INERTIA_START
            + (
                PSO_INERTIA_END
                - PSO_INERTIA_START
            )
            * progress
        )

        r1 = rng.random(
            size=(
                PSO_SWARM_SIZE,
                N_COEFFICIENTS
            )
        )

        r2 = rng.random(
            size=(
                PSO_SWARM_SIZE,
                N_COEFFICIENTS
            )
        )

        cognitive_component = (
            PSO_COGNITIVE
            * r1
            * (
                personal_best_positions
                - positions
            )
        )

        social_component = (
            PSO_SOCIAL
            * r2
            * (
                global_best_position
                - positions
            )
        )

        velocities = (
            inertia * velocities
            + cognitive_component
            + social_component
        )

        velocities = np.clip(
            velocities,
            -velocity_limit,
            velocity_limit
        )

        positions = (
            positions
            + velocities
        )

        below_lower = (
            positions
            < lower_bounds
        )

        above_upper = (
            positions
            > upper_bounds
        )

        velocities[
            below_lower | above_upper
        ] *= -0.5

        positions = np.clip(
            positions,
            lower_bounds,
            upper_bounds
        )

        current_scores = np.array([
            rmse_objective(position)
            for position in positions
        ])

        improved = (
            current_scores
            < personal_best_scores
        )

        personal_best_positions[
            improved
        ] = positions[
            improved
        ]

        personal_best_scores[
            improved
        ] = current_scores[
            improved
        ]

        candidate_index = np.argmin(
            personal_best_scores
        )

        candidate_score = float(
            personal_best_scores[
                candidate_index
            ]
        )

        if (
            candidate_score
            < global_best_score
        ):

            global_best_score = (
                candidate_score
            )

            global_best_position = (
                personal_best_positions[
                    candidate_index
                ].copy()
            )

        best_history.append(
            global_best_score
        )

    return (
        global_best_position,
        global_best_score,
        np.asarray(best_history)
    )


# ============================================================
# 9. MULTIPLE INDEPENDENT RUNS
# ============================================================

ga_runs = []
pso_runs = []

ga_histories = []
pso_histories = []

print(
    "\nRunning independent GA optimizations..."
)

for run in range(N_RUNS):

    seed = (
        BASE_RANDOM_SEED
        + run
    )

    beta, rmse, history = run_ga(
        seed=seed
    )

    ga_runs.append({
        "Run": run + 1,
        "Seed": seed,
        "RMSE": rmse,
        "Beta": beta
    })

    ga_histories.append(
        history
    )

    print(
        f"GA run {run + 1:02d}/{N_RUNS} | "
        f"RMSE = {rmse:.8f}"
    )


print(
    "\nRunning independent PSO optimizations..."
)

for run in range(N_RUNS):

    seed = (
        BASE_RANDOM_SEED
        + 1000
        + run
    )

    beta, rmse, history = run_pso(
        seed=seed
    )

    pso_runs.append({
        "Run": run + 1,
        "Seed": seed,
        "RMSE": rmse,
        "Beta": beta
    })

    pso_histories.append(
        history
    )

    print(
        f"PSO run {run + 1:02d}/{N_RUNS} | "
        f"RMSE = {rmse:.8f}"
    )


# ============================================================
# 10. SELECT BEST RUN OF EACH METHOD
# ============================================================

best_ga_run = min(
    ga_runs,
    key=lambda item: item["RMSE"]
)

best_pso_run = min(
    pso_runs,
    key=lambda item: item["RMSE"]
)

beta_ga = best_ga_run[
    "Beta"
]

beta_pso = best_pso_run[
    "Beta"
]

metrics_ga = calculate_metrics(
    beta_ga
)

metrics_pso = calculate_metrics(
    beta_pso
)

pred_ga = predict_ai(
    beta_ga
)

pred_pso = predict_ai(
    beta_pso
)


# ============================================================
# 11. COEFFICIENTS AND PERFORMANCE
# ============================================================

coefficient_table = pd.DataFrame({
    "Coefficient": coefficient_names,
    "GA": beta_ga,
    "PSO": beta_pso
})

metrics_table = pd.DataFrame({
    "Metric": [
        "R2",
        "RMSE",
        "MAE",
        "MAPE (%)"
    ],
    "GA": [
        metrics_ga["R2"],
        metrics_ga["RMSE"],
        metrics_ga["MAE"],
        metrics_ga["MAPE (%)"]
    ],
    "PSO": [
        metrics_pso["R2"],
        metrics_pso["RMSE"],
        metrics_pso["MAE"],
        metrics_pso["MAPE (%)"]
    ]
})

print(
    "\nBest-fit coefficients:"
)

display(
    coefficient_table.round(8)
)

print(
    "\nModel performance:"
)

display(
    metrics_table.round(8)
)


# ============================================================
# 12. MULTI-RUN STABILITY
# ============================================================

ga_rmse_values = np.array([
    run["RMSE"]
    for run in ga_runs
])

pso_rmse_values = np.array([
    run["RMSE"]
    for run in pso_runs
])

run_summary_table = pd.DataFrame({
    "Method": [
        "GA",
        "PSO"
    ],
    "Number of runs": [
        N_RUNS,
        N_RUNS
    ],
    "Best RMSE": [
        ga_rmse_values.min(),
        pso_rmse_values.min()
    ],
    "Mean RMSE": [
        ga_rmse_values.mean(),
        pso_rmse_values.mean()
    ],
    "RMSE standard deviation": [
        ga_rmse_values.std(ddof=1),
        pso_rmse_values.std(ddof=1)
    ],
    "Worst RMSE": [
        ga_rmse_values.max(),
        pso_rmse_values.max()
    ]
})

print(
    "\nMulti-run stability:"
)

display(
    run_summary_table.round(8)
)


# ============================================================
# 13. FINAL EMPIRICAL EQUATIONS
# ============================================================

def signed_term(
    coefficient,
    expression
):

    sign = (
        "+"
        if coefficient >= 0
        else "-"
    )

    return (
        f" {sign} "
        f"{abs(coefficient):.8f}"
        f"{expression}"
    )


def build_equation(
    beta,
    method
):

    equation = (
        f"AI_{method} = "
        f"{beta[0]:.8f}"
    )

    equation += signed_term(
        beta[1],
        "(f/b)"
    )

    equation += signed_term(
        beta[2],
        "log10(w)"
    )

    equation += signed_term(
        beta[3],
        "(f/b)^2"
    )

    equation += signed_term(
        beta[4],
        "[log10(w)]^2"
    )

    equation += signed_term(
        beta[5],
        "(f/b)log10(w)"
    )

    equation += signed_term(
        beta[6],
        "D_cement"
    )

    equation += signed_term(
        beta[7],
        "D_chromium"
    )

    return equation


print(
    "\nFinal GA equation:"
)

print(
    build_equation(
        beta_ga,
        "GA"
    )
)

print(
    "\nFinal PSO equation:"
)

print(
    build_equation(
        beta_pso,
        "PSO"
    )
)


# ============================================================
# 14. CONVERGENCE PLOT
# ============================================================

best_ga_history = ga_histories[
    best_ga_run["Run"] - 1
]

best_pso_history = pso_histories[
    best_pso_run["Run"] - 1
]

plt.figure(
    figsize=(8, 5)
)

plt.plot(
    np.arange(
        1,
        len(best_ga_history) + 1
    ),
    best_ga_history,
    label="GA"
)

plt.plot(
    np.arange(
        1,
        len(best_pso_history) + 1
    ),
    best_pso_history,
    label="PSO"
)

plt.xlabel(
    "Generation / iteration"
)

plt.ylabel(
    "Best RMSE"
)

plt.title(
    "GA and PSO Convergence"
)

plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()

plt.show()


# ============================================================
# 15. MEASURED VERSUS PREDICTED AGING INDEX
# ============================================================

minimum_value = min(
    y.min(),
    pred_ga.min(),
    pred_pso.min()
)

maximum_value = max(
    y.max(),
    pred_ga.max(),
    pred_pso.max()
)

plt.figure(
    figsize=(7, 6)
)

plt.scatter(
    y,
    pred_ga,
    label="GA",
    s=55,
    alpha=0.8
)

plt.scatter(
    y,
    pred_pso,
    label="PSO",
    marker="s",
    s=55,
    alpha=0.8
)

plt.plot(
    [
        minimum_value,
        maximum_value
    ],
    [
        minimum_value,
        maximum_value
    ],
    linestyle="--",
    linewidth=1.5,
    label="Ideal agreement"
)

plt.xlabel(
    "Measured Aging Index"
)

plt.ylabel(
    "Predicted Aging Index"
)

plt.title(
    "Measured versus Predicted Aging Index"
)

plt.legend()
plt.grid(alpha=0.25)
plt.tight_layout()

plt.show()


# ============================================================
# 16. PREDICTION TABLE
# ============================================================

prediction_table = data.copy()

prediction_table[
    "log10_frequency"
] = log_frequency

prediction_table[
    "D_cement"
] = D_cement

prediction_table[
    "D_chromium"
] = D_chromium

prediction_table[
    "AI_GA_predicted"
] = pred_ga

prediction_table[
    "AI_PSO_predicted"
] = pred_pso

prediction_table[
    "GA_residual"
] = y - pred_ga

prediction_table[
    "PSO_residual"
] = y - pred_pso

print(
    "\nPredictions:"
)

display(
    prediction_table.round(8)
)


# ============================================================
# 17. SAVE RESULTS
# ============================================================

OUTPUT_FILE = (
    "/content/GA_PSO_Aging_Index_Independent_Results.xlsx"
)

with pd.ExcelWriter(
    OUTPUT_FILE,
    engine="openpyxl"
) as writer:

    coefficient_table.to_excel(
        writer,
        sheet_name="Best Coefficients",
        index=False
    )

    metrics_table.to_excel(
        writer,
        sheet_name="Best Metrics",
        index=False
    )

    run_summary_table.to_excel(
        writer,
        sheet_name="Run Summary",
        index=False
    )

    prediction_table.to_excel(
        writer,
        sheet_name="Predictions",
        index=False
    )

print(
    "\nResults saved to:"
)

print(
    OUTPUT_FILE
)
```
