```python
# ============================================================
# ANN Prediction of Aging Index
# Nested Leave-One-Mastic-Out Cross-Validation
#
# Input file:
#   /content/Aging index.xlsx
#
# Inputs:
#   1. Filler type
#   2. Mass-based filler-to-bitumen ratio
#   3. log10(Frequency)
#
# Output:
#   Aging Index
# ============================================================

import warnings
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import LeaveOneGroupOut, GridSearchCV
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer, TransformedTargetRegressor
from sklearn.pipeline import Pipeline
from sklearn.neural_network import MLPRegressor
from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error
)

warnings.filterwarnings("ignore")

# ============================================================
# 1. Read the Excel file
# ============================================================

file_path = "/content/Aging index.xlsx"

data = pd.read_excel(file_path)

data.columns = (
    data.columns
    .astype(str)
    .str.strip()
)

print("Original columns:")
print(data.columns.tolist())


# ============================================================
# 2. Rename possible column names
# ============================================================

rename_map = {
    # Filler type
    "Filler type": "Filler_type",
    "Filler Type": "Filler_type",
    "Filler_type": "Filler_type",

    # Mass-based F/B ratio
    "F/B mass": "FB_mass",
    "FB mass": "FB_mass",
    "FB_mass": "FB_mass",
    "f/b by weight": "FB_mass",
    "F/B by weight": "FB_mass",
    "F/B By Weight": "FB_mass",

    # Frequency
    "Frequency (rad/s)": "Frequency",
    "Frequency": "Frequency",
    "Frequency (rad/sec)": "Frequency",
    "Reduced angular frequency (rad/s)": "Frequency",

    # Aging Index
    "Aging Index": "Aging_Index",
    "Aging index": "Aging_Index",
    "Aging_Index": "Aging_Index",
    "AI": "Aging_Index",
    "AI @ freq.": "Aging_Index"
}

data = data.rename(columns=rename_map)


# ============================================================
# 3. Automatic column detection
# ============================================================

def find_column(df, accepted_names, keywords):
    """
    Detect a column using exact accepted names first,
    followed by keyword matching.
    """

    for name in accepted_names:
        if name in df.columns:
            return name

    for col in df.columns:
        col_lower = str(col).lower()

        if all(
            keyword.lower() in col_lower
            for keyword in keywords
        ):
            return col

    return None


filler_col = find_column(
    data,
    accepted_names=["Filler_type"],
    keywords=["filler"]
)

fbmass_col = find_column(
    data,
    accepted_names=["FB_mass"],
    keywords=["weight"]
)

if fbmass_col is None:
    fbmass_col = find_column(
        data,
        accepted_names=["FB_mass"],
        keywords=["mass"]
    )

frequency_col = find_column(
    data,
    accepted_names=["Frequency"],
    keywords=["frequency"]
)

aging_col = find_column(
    data,
    accepted_names=["Aging_Index"],
    keywords=["aging", "index"]
)

if aging_col is None:
    aging_col = find_column(
        data,
        accepted_names=["Aging_Index"],
        keywords=["ai"]
    )

detected_columns = {
    "Filler_type": filler_col,
    "FB_mass": fbmass_col,
    "Frequency": frequency_col,
    "Aging_Index": aging_col
}

print("\nDetected columns:")
print(detected_columns)

missing_columns = [
    standard_name
    for standard_name, detected_name
    in detected_columns.items()
    if detected_name is None
]

if missing_columns:
    raise ValueError(
        "The following required columns could not be detected: "
        f"{missing_columns}\n"
        f"Available columns: {data.columns.tolist()}"
    )

data = data.rename(
    columns={
        filler_col: "Filler_type",
        fbmass_col: "FB_mass",
        frequency_col: "Frequency",
        aging_col: "Aging_Index"
    }
)


# ============================================================
# 4. Keep and clean the required columns
# ============================================================

required_columns = [
    "Filler_type",
    "FB_mass",
    "Frequency",
    "Aging_Index"
]

data = data[required_columns].copy()

data["Filler_type"] = (
    data["Filler_type"]
    .astype(str)
    .str.strip()
)

numeric_columns = [
    "FB_mass",
    "Frequency",
    "Aging_Index"
]

for column in numeric_columns:
    data[column] = pd.to_numeric(
        data[column],
        errors="coerce"
    )

data = (
    data
    .dropna(subset=required_columns)
    .reset_index(drop=True)
)

data = data[
    (data["Frequency"] > 0) &
    (data["Aging_Index"] > 0)
].reset_index(drop=True)


# ============================================================
# 5. Normalize filler names
# ============================================================

filler_name_map = {
    "Rh": "Rheofiller",
    "RH": "Rheofiller",
    "rh": "Rheofiller",
    "rheofiller": "Rheofiller",
    "RHEOFILLER": "Rheofiller",

    "Ce": "Cement",
    "CE": "Cement",
    "ce": "Cement",
    "cement": "Cement",
    "CEMENT": "Cement",

    "Ch": "Chromium steel sludge",
    "CH": "Chromium steel sludge",
    "ch": "Chromium steel sludge",
    "Chromium Steel Sludge": "Chromium steel sludge",
    "chromium steel sludge": "Chromium steel sludge"
}

data["Filler_type"] = (
    data["Filler_type"]
    .replace(filler_name_map)
)

print("\nFiller types:")
print(data["Filler_type"].unique())


# ============================================================
# 6. Transform frequency
# ============================================================

data["log10_Frequency"] = np.log10(
    data["Frequency"]
)


# ============================================================
# 7. Create mastic groups
#
# Each mastic is defined by:
#   Filler type + mass-based F/B ratio
#
# All four frequency points from one mastic are removed
# together during each outer validation fold.
# ============================================================

data["Mastic_Group"] = (
    data["Filler_type"].astype(str)
    + "_FBmass_"
    + data["FB_mass"].round(3).astype(str)
)

print("\nClean dataset:")
display(data)

print("\nNumber of observations:")
print(len(data))

print("\nNumber of independent mastic groups:")
print(data["Mastic_Group"].nunique())

group_size_check = (
    data
    .groupby("Mastic_Group")
    .size()
    .reset_index(name="Number_of_rows")
)

print("\nNumber of rows in each mastic group:")
display(group_size_check)

if len(data) != 36:
    print(
        "\nWarning: The expected dataset contains 36 observations, "
        f"but {len(data)} observations were detected."
    )

if data["Mastic_Group"].nunique() != 9:
    print(
        "\nWarning: Nine mastic groups were expected, but "
        f"{data['Mastic_Group'].nunique()} groups were detected."
    )


# ============================================================
# 8. Define inputs, output, and groups
# ============================================================

input_columns = [
    "Filler_type",
    "FB_mass",
    "log10_Frequency"
]

X = data[input_columns].copy()

y = data["Aging_Index"].to_numpy()

groups = data["Mastic_Group"].to_numpy()


# ============================================================
# 9. Input preprocessing
# ============================================================

categorical_features = [
    "Filler_type"
]

numerical_features = [
    "FB_mass",
    "log10_Frequency"
]

filler_categories = [[
    "Chromium steel sludge",
    "Cement",
    "Rheofiller"
]]

try:
    one_hot_encoder = OneHotEncoder(
        categories=filler_categories,
        handle_unknown="ignore",
        sparse_output=False
    )

except TypeError:
    # Compatibility with older versions of scikit-learn
    one_hot_encoder = OneHotEncoder(
        categories=filler_categories,
        handle_unknown="ignore",
        sparse=False
    )

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            one_hot_encoder,
            categorical_features
        ),
        (
            "numerical",
            StandardScaler(),
            numerical_features
        )
    ],
    remainder="drop"
)


# ============================================================
# 10. ANN model
# ============================================================

def make_ann_model():
    """
    Create an ANN model with input preprocessing
    and target standardization.
    """

    ann = MLPRegressor(
        random_state=42,
        max_iter=10000,
        early_stopping=False
    )

    pipeline = Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            ("ann", ann)
        ]
    )

    model = TransformedTargetRegressor(
        regressor=pipeline,
        transformer=StandardScaler()
    )

    return model


# ============================================================
# 11. ANN hyperparameter grid
#
# Exact hyperparameters selected for this analysis
# ============================================================

param_grid = {
    "regressor__ann__hidden_layer_sizes": [
        (4,),
        (6,),
        (8,),
        (10,),
        (12,)
    ],

    "regressor__ann__activation": [
        "relu",
        "tanh"
    ],

    "regressor__ann__solver": [
        "lbfgs",
        "adam"
    ],

    "regressor__ann__alpha": [
        0.0001,
        0.001,
        0.01,
        0.1
    ]
}

number_of_combinations = (
    len(param_grid["regressor__ann__hidden_layer_sizes"])
    * len(param_grid["regressor__ann__activation"])
    * len(param_grid["regressor__ann__solver"])
    * len(param_grid["regressor__ann__alpha"])
)

print("\nNumber of ANN hyperparameter combinations:")
print(number_of_combinations)


# ============================================================
# 12. Nested Leave-One-Mastic-Out cross-validation
#
# Outer loop:
#   Evaluates the model on one completely unseen mastic.
#
# Inner loop:
#   Selects ANN hyperparameters using only the training data.
# ============================================================

outer_cv = LeaveOneGroupOut()

y_true_all = []
y_pred_all = []

test_indices_all = []
heldout_groups_all = []
best_parameters_all = []

fold_summary_rows = []

for fold_number, (train_idx, test_idx) in enumerate(
    outer_cv.split(X, y, groups),
    start=1
):

    X_train = X.iloc[train_idx].copy()
    X_test = X.iloc[test_idx].copy()

    y_train = y[train_idx]
    y_test = y[test_idx]

    groups_train = groups[train_idx]

    heldout_group = np.unique(
        groups[test_idx]
    )[0]

    print(
        f"\nOuter fold {fold_number}: "
        f"holding out {heldout_group}"
    )

    ann_model = make_ann_model()

    inner_cv = LeaveOneGroupOut()

    grid_search = GridSearchCV(
        estimator=ann_model,
        param_grid=param_grid,
        scoring="neg_root_mean_squared_error",
        cv=inner_cv,
        n_jobs=-1,
        refit=True,
        return_train_score=True,
        error_score="raise"
    )

    grid_search.fit(
        X_train,
        y_train,
        groups=groups_train
    )

    best_model_fold = grid_search.best_estimator_

    y_pred_fold = best_model_fold.predict(
        X_test
    )

    y_true_all.extend(y_test.tolist())
    y_pred_all.extend(y_pred_fold.tolist())

    test_indices_all.extend(test_idx.tolist())

    heldout_groups_all.extend(
        [heldout_group] * len(test_idx)
    )

    best_parameters_string = str(
        grid_search.best_params_
    )

    best_parameters_all.extend(
        [best_parameters_string] * len(test_idx)
    )

    fold_rmse = np.sqrt(
        mean_squared_error(
            y_test,
            y_pred_fold
        )
    )

    fold_mae = mean_absolute_error(
        y_test,
        y_pred_fold
    )

    fold_summary_rows.append({
        "Outer_Fold": fold_number,
        "Heldout_Group": heldout_group,
        "Number_of_Test_Observations": len(test_idx),
        "Fold_RMSE": fold_rmse,
        "Fold_MAE": fold_mae,
        "Best_Inner_CV_RMSE": -grid_search.best_score_,
        "Best_Parameters": best_parameters_string
    })

    print("Best parameters:")
    print(grid_search.best_params_)

    print("Inner LOMO RMSE:")
    print(-grid_search.best_score_)

    print("Outer-fold RMSE:")
    print(fold_rmse)


# ============================================================
# 13. Restore the original row order
# ============================================================

y_true_all = np.asarray(
    y_true_all,
    dtype=float
)

y_pred_all = np.asarray(
    y_pred_all,
    dtype=float
)

test_indices_all = np.asarray(
    test_indices_all,
    dtype=int
)

heldout_groups_all = np.asarray(
    heldout_groups_all,
    dtype=object
)

best_parameters_all = np.asarray(
    best_parameters_all,
    dtype=object
)

original_order = np.argsort(
    test_indices_all
)

y_true_all = y_true_all[
    original_order
]

y_pred_all = y_pred_all[
    original_order
]

test_indices_all = test_indices_all[
    original_order
]

heldout_groups_all = heldout_groups_all[
    original_order
]

best_parameters_all = best_parameters_all[
    original_order
]


# ============================================================
# 14. Evaluation functions
# ============================================================

def calculate_rmse(y_true, y_pred):
    return np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )


def calculate_mape(y_true, y_pred):
    y_true = np.asarray(
        y_true,
        dtype=float
    )

    y_pred = np.asarray(
        y_pred,
        dtype=float
    )

    valid = y_true != 0

    if not np.any(valid):
        return np.nan

    return (
        np.mean(
            np.abs(
                (
                    y_true[valid]
                    - y_pred[valid]
                )
                / y_true[valid]
            )
        )
        * 100
    )


def calculate_smape(y_true, y_pred):
    y_true = np.asarray(
        y_true,
        dtype=float
    )

    y_pred = np.asarray(
        y_pred,
        dtype=float
    )

    denominator = (
        np.abs(y_true)
        + np.abs(y_pred)
    )

    valid = denominator != 0

    if not np.any(valid):
        return np.nan

    return (
        np.mean(
            2
            * np.abs(
                y_pred[valid]
                - y_true[valid]
            )
            / denominator[valid]
        )
        * 100
    )


# ============================================================
# 15. Overall nested LOMO metrics
# ============================================================

overall_r2 = r2_score(
    y_true_all,
    y_pred_all
)

overall_rmse = calculate_rmse(
    y_true_all,
    y_pred_all
)

overall_mae = mean_absolute_error(
    y_true_all,
    y_pred_all
)

overall_mape = calculate_mape(
    y_true_all,
    y_pred_all
)

overall_smape = calculate_smape(
    y_true_all,
    y_pred_all
)

metrics = pd.DataFrame([
    {
        "Validation": "Nested Leave-One-Mastic-Out",
        "Model": "ANN",
        "Target": "Aging Index",
        "R2": overall_r2,
        "RMSE": overall_rmse,
        "MAE": overall_mae,
        "MAPE_%": overall_mape,
        "sMAPE_%": overall_smape
    }
])

print(
    "\nANN nested Leave-One-Mastic-Out metrics:"
)

display(metrics)


# ============================================================
# 16. Create prediction table
# ============================================================

predictions = (
    data
    .iloc[test_indices_all]
    .copy()
    .reset_index(drop=True)
)

predictions["Heldout_Group"] = (
    heldout_groups_all
)

predictions["Actual_Aging_Index"] = (
    y_true_all
)

predictions["Predicted_Aging_Index"] = (
    y_pred_all
)

predictions["Residual"] = (
    predictions["Actual_Aging_Index"]
    - predictions["Predicted_Aging_Index"]
)

predictions["Absolute_Error"] = np.abs(
    predictions["Residual"]
)

predictions["Percentage_Error_%"] = (
    predictions["Residual"]
    / predictions["Actual_Aging_Index"]
) * 100

predictions["Absolute_Percentage_Error_%"] = np.abs(
    predictions["Percentage_Error_%"]
)

predictions["Best_Parameters_in_Outer_Fold"] = (
    best_parameters_all
)

print("\nNested LOMO predictions:")
display(predictions)


# ============================================================
# 17. Outer-fold summary
# ============================================================

fold_summary = pd.DataFrame(
    fold_summary_rows
)

print("\nOuter-fold summary:")
display(fold_summary)


# ============================================================
# 18. Metrics for each held-out mastic
# ============================================================

group_metric_rows = []

for group_name, group_df in predictions.groupby(
    "Mastic_Group"
):

    group_y_true = group_df[
        "Actual_Aging_Index"
    ].to_numpy()

    group_y_pred = group_df[
        "Predicted_Aging_Index"
    ].to_numpy()

    group_metric_rows.append({
        "Mastic_Group": group_name,

        "Filler_type":
            group_df["Filler_type"].iloc[0],

        "FB_mass":
            group_df["FB_mass"].iloc[0],

        "R2":
            r2_score(
                group_y_true,
                group_y_pred
            ),

        "RMSE":
            calculate_rmse(
                group_y_true,
                group_y_pred
            ),

        "MAE":
            mean_absolute_error(
                group_y_true,
                group_y_pred
            ),

        "MAPE_%":
            calculate_mape(
                group_y_true,
                group_y_pred
            ),

        "sMAPE_%":
            calculate_smape(
                group_y_true,
                group_y_pred
            )
    })

group_metrics = pd.DataFrame(
    group_metric_rows
)

print("\nPerformance for each held-out mastic:")
display(group_metrics)


# ============================================================
# 19. Actual versus predicted plot
# ============================================================

plot_min = min(
    y_true_all.min(),
    y_pred_all.min()
)

plot_max = max(
    y_true_all.max(),
    y_pred_all.max()
)

plot_range = plot_max - plot_min

if plot_range == 0:
    plot_range = 1

margin = 0.05 * plot_range

parity_min = plot_min - margin
parity_max = plot_max + margin

plt.figure(figsize=(6, 6))

plt.scatter(
    y_true_all,
    y_pred_all,
    s=55
)

plt.plot(
    [parity_min, parity_max],
    [parity_min, parity_max],
    linewidth=1.5,
    label="Perfect prediction"
)

plt.xlabel("Experimental Aging Index")
plt.ylabel("Predicted Aging Index")

plt.title(
    "ANN Prediction of Aging Index\n"
    "Nested Leave-One-Mastic-Out Validation"
)

plt.xlim(
    parity_min,
    parity_max
)

plt.ylim(
    parity_min,
    parity_max
)

plt.grid(
    True,
    alpha=0.35
)

plt.legend()
plt.tight_layout()

plt.savefig(
    "ANN_AgingIndex_LOMO_Actual_vs_Predicted.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 20. Residual plot
# ============================================================

residuals = (
    y_true_all
    - y_pred_all
)

plt.figure(figsize=(6, 4))

plt.scatter(
    y_pred_all,
    residuals,
    s=55
)

plt.axhline(
    y=0,
    linewidth=1.5
)

plt.xlabel("Predicted Aging Index")
plt.ylabel("Residual")

plt.title(
    "ANN Residuals\n"
    "Nested Leave-One-Mastic-Out Validation"
)

plt.grid(
    True,
    alpha=0.35
)

plt.tight_layout()

plt.savefig(
    "ANN_AgingIndex_LOMO_Residuals.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 21. Experimental and predicted AI versus frequency
# ============================================================

curve_data = predictions.sort_values(
    by=[
        "Filler_type",
        "FB_mass",
        "Frequency"
    ]
)

for (
    filler_type,
    fb_mass
), subset in curve_data.groupby(
    [
        "Filler_type",
        "FB_mass"
    ]
):

    subset = subset.sort_values(
        "Frequency"
    )

    plt.figure(figsize=(6, 4))

    plt.plot(
        subset["Frequency"],
        subset["Actual_Aging_Index"],
        marker="o",
        label="Experimental"
    )

    plt.plot(
        subset["Frequency"],
        subset["Predicted_Aging_Index"],
        marker="s",
        linestyle="--",
        label="ANN prediction"
    )

    plt.xscale("log")

    plt.xlabel(
        "Reduced angular frequency (rad/s)"
    )

    plt.ylabel(
        "Aging Index"
    )

    plt.title(
        f"{filler_type}, F/B mass = {fb_mass}"
    )

    plt.grid(
        True,
        which="both",
        alpha=0.35
    )

    plt.legend()
    plt.tight_layout()

    safe_filler_name = (
        filler_type
        .replace(" ", "_")
        .replace("/", "_")
    )

    plt.savefig(
        f"ANN_AI_Curve_{safe_filler_name}_"
        f"FBmass_{fb_mass}.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ============================================================
# 22. Train the final ANN using all observations
#
# Hyperparameters are selected using grouped LOMO CV.
# This final model is used for future prediction, not for
# reporting the nested-CV performance.
# ============================================================

final_ann_model = make_ann_model()

final_grid_search = GridSearchCV(
    estimator=final_ann_model,
    param_grid=param_grid,
    scoring="neg_root_mean_squared_error",
    cv=LeaveOneGroupOut(),
    n_jobs=-1,
    refit=True,
    return_train_score=True,
    error_score="raise"
)

final_grid_search.fit(
    X,
    y,
    groups=groups
)

best_final_model = (
    final_grid_search.best_estimator_
)

best_final_parameters = (
    final_grid_search.best_params_
)

print("\nBest final ANN parameters:")
print(best_final_parameters)

print("\nBest grouped-CV RMSE for the final model:")
print(-final_grid_search.best_score_)


# ============================================================
# 23. Final grid-search results
# ============================================================

grid_results = pd.DataFrame(
    final_grid_search.cv_results_
)

grid_result_columns = [
    "rank_test_score",
    "mean_test_score",
    "std_test_score",
    "mean_train_score",
    "std_train_score",
    "param_regressor__ann__hidden_layer_sizes",
    "param_regressor__ann__activation",
    "param_regressor__ann__solver",
    "param_regressor__ann__alpha"
]

grid_results_clean = (
    grid_results[grid_result_columns]
    .copy()
    .sort_values("rank_test_score")
)

grid_results_clean[
    "Grouped_CV_RMSE"
] = -grid_results_clean[
    "mean_test_score"
]

grid_results_clean[
    "Grouped_Training_RMSE"
] = -grid_results_clean[
    "mean_train_score"
]

print("\nFinal ANN grid-search results:")
display(grid_results_clean)


# ============================================================
# 24. Save the final trained ANN
# ============================================================

ann_framework = {
    "model": best_final_model,

    "input_columns": input_columns,

    "target": "Aging_Index",

    "best_parameters": best_final_parameters,

    "input_description": (
        "Filler type, mass-based filler-to-bitumen ratio, "
        "and log10 of reduced angular frequency"
    ),

    "validation": (
        "Nested Leave-One-Mastic-Out cross-validation"
    ),

    "number_of_observations": len(data),

    "number_of_mastic_groups":
        data["Mastic_Group"].nunique()
}

model_file = (
    "ANN_AgingIndex_FBmass_LOMO_Final.pkl"
)

joblib.dump(
    ann_framework,
    model_file
)


# ============================================================
# 25. Export all numerical results to Excel
# ============================================================

parity_line_data = pd.DataFrame({
    "Parity_x": [
        parity_min,
        parity_max
    ],

    "Parity_y": [
        parity_min,
        parity_max
    ]
})

plot_data = pd.DataFrame({
    "Experimental_Aging_Index":
        y_true_all,

    "Predicted_Aging_Index":
        y_pred_all,

    "Residual":
        residuals
})

best_final_parameters_table = pd.DataFrame(
    [
        {
            "Parameter": parameter,
            "Selected_Value": str(value)
        }
        for parameter, value
        in best_final_parameters.items()
    ]
)

output_file = (
    "ANN_AgingIndex_FBmass_LOMO_Results.xlsx"
)

with pd.ExcelWriter(
    output_file,
    engine="openpyxl"
) as writer:

    data.to_excel(
        writer,
        sheet_name="Clean_Data",
        index=False
    )

    metrics.to_excel(
        writer,
        sheet_name="Overall_Metrics",
        index=False
    )

    predictions.to_excel(
        writer,
        sheet_name="LOMO_Predictions",
        index=False
    )

    fold_summary.to_excel(
        writer,
        sheet_name="Outer_Fold_Summary",
        index=False
    )

    group_metrics.to_excel(
        writer,
        sheet_name="Mastic_Metrics",
        index=False
    )

    grid_results_clean.to_excel(
        writer,
        sheet_name="Final_Grid_Search",
        index=False
    )

    best_final_parameters_table.to_excel(
        writer,
        sheet_name="Best_Final_Parameters",
        index=False
    )

    plot_data.to_excel(
        writer,
        sheet_name="Plot_Data",
        index=False
    )

    parity_line_data.to_excel(
        writer,
        sheet_name="Parity_Line",
        index=False
    )


# ============================================================
# 26. Final output summary
# ============================================================

print("\nAnalysis completed successfully.")

print("\nOverall nested LOMO performance:")
print(f"R2     = {overall_r2:.6f}")
print(f"RMSE   = {overall_rmse:.6f}")
print(f"MAE    = {overall_mae:.6f}")
print(f"MAPE   = {overall_mape:.2f}%")
print(f"sMAPE  = {overall_smape:.2f}%")

print("\nSaved files:")
print(
    "1. ANN_AgingIndex_FBmass_LOMO_Results.xlsx"
)
print(
    "2. ANN_AgingIndex_FBmass_LOMO_Final.pkl"
)
print(
    "3. ANN_AgingIndex_LOMO_Actual_vs_Predicted.png"
)
print(
    "4. ANN_AgingIndex_LOMO_Residuals.png"
)
print(
    "5. Nine experimental-versus-predicted curve figures"
)
```
