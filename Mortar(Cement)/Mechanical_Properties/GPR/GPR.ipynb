```python
# ============================================================
# Publication-ready GPR framework
# Two independent GPR models inside one framework
# Input file: Final.xlsx
# Inputs: Sludge %, Plasticizer %, Density
# Outputs:
#   GPR-1 -> sigma_m
#   GPR-2 -> sigma_c_mean
# Validation: LOOCV
# Kernel tuning: RBF, Matern, RationalQuadratic
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings
import joblib

from sklearn.model_selection import LeaveOneOut, GridSearchCV, cross_val_predict
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.compose import TransformedTargetRegressor
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import (
    RBF,
    Matern,
    RationalQuadratic,
    ConstantKernel as C,
    WhiteKernel
)
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

warnings.filterwarnings("ignore")

# ------------------------------------------------------------
# 1. Import Excel file
# ------------------------------------------------------------

data = pd.read_excel("Final.xlsx")
data.columns = data.columns.astype(str).str.strip()

rename_map = {
    "Sample ID": "Sample_ID",
    "Sample_ID": "Sample_ID",

    "Sludge %": "Sludge_%",
    "Sludge_%": "Sludge_%",

    "Plasticizer %": "Plasticizer_%",
    "Plasticizer_%": "Plasticizer_%",

    "Density": "Density",

    "σm Flexural (MPa)": "sigma_m_MPa",
    "sigma_m_MPa": "sigma_m_MPa",

    "mean σc Compression (MPa)": "sigma_c_mean_MPa",
    "sigma_c_mean_MPa": "sigma_c_mean_MPa"
}

data = data.rename(columns=rename_map)

print("Columns in file:")
print(data.columns.tolist())

# ------------------------------------------------------------
# 2. Define inputs and outputs
# ------------------------------------------------------------

X_cols = ["Sludge_%", "Plasticizer_%", "Density"]
y_cols = ["sigma_m_MPa", "sigma_c_mean_MPa"]

required_cols = ["Sample_ID"] + X_cols + y_cols
missing_cols = [col for col in required_cols if col not in data.columns]

if len(missing_cols) > 0:
    raise ValueError(f"Missing columns in Excel file: {missing_cols}")

data = data[required_cols].dropna().reset_index(drop=True)

X = data[X_cols].values

display(data)

# ------------------------------------------------------------
# 3. Define kernels for tuning
# ------------------------------------------------------------

n_features = len(X_cols)

kernel_options = [
    C(1.0, (1e-3, 1e3))
    * RBF(length_scale=np.ones(n_features), length_scale_bounds=(1e-3, 1e3))
    + WhiteKernel(noise_level=1e-3, noise_level_bounds=(1e-8, 1e1)),

    C(1.0, (1e-3, 1e3))
    * Matern(length_scale=np.ones(n_features), length_scale_bounds=(1e-3, 1e3), nu=1.5)
    + WhiteKernel(noise_level=1e-3, noise_level_bounds=(1e-8, 1e1)),

    C(1.0, (1e-3, 1e3))
    * RationalQuadratic(length_scale=1.0, alpha=1.0)
    + WhiteKernel(noise_level=1e-3, noise_level_bounds=(1e-8, 1e1))
]

param_grid = {
    "regressor__gpr__kernel": kernel_options
}

loo = LeaveOneOut()

# ------------------------------------------------------------
# 4. Helper functions
# ------------------------------------------------------------

def mape(y_true, y_pred):
    return np.mean(np.abs((y_true - y_pred) / y_true)) * 100


def run_single_target_gpr(target_col, target_label):
    print("\n" + "=" * 70)
    print(f"Running GPR for: {target_label}")
    print("=" * 70)

    y = data[target_col].values

    base_gpr = GaussianProcessRegressor(
        normalize_y=False,
        n_restarts_optimizer=10,
        random_state=42
    )

    pipe = Pipeline([
        ("x_scaler", StandardScaler()),
        ("gpr", base_gpr)
    ])

    model = TransformedTargetRegressor(
        regressor=pipe,
        transformer=StandardScaler()
    )

    grid = GridSearchCV(
        estimator=model,
        param_grid=param_grid,
        scoring="neg_root_mean_squared_error",
        cv=loo,
        n_jobs=-1,
        return_train_score=True
    )

    grid.fit(X, y)

    best_model = grid.best_estimator_
    best_params = grid.best_params_

    print("\nBest kernel:")
    print(best_params["regressor__gpr__kernel"])

    # LOOCV predictions using selected kernel
    y_pred_loocv = cross_val_predict(
        best_model,
        X,
        y,
        cv=loo,
        n_jobs=-1
    )

    # Metrics
    r2 = r2_score(y, y_pred_loocv)
    mae = mean_absolute_error(y, y_pred_loocv)
    rmse = np.sqrt(mean_squared_error(y, y_pred_loocv))
    mape_value = mape(y, y_pred_loocv)

    metrics = {
        "Target": target_col,
        "R2": r2,
        "RMSE": rmse,
        "MAE": mae,
        "MAPE_%": mape_value,
        "Best_kernel": str(best_params["regressor__gpr__kernel"])
    }

    metrics_df = pd.DataFrame([metrics])

    print("\nLOOCV metrics:")
    display(metrics_df)

    # Prediction table
    predictions_df = data[["Sample_ID"] + X_cols + [target_col]].copy()
    predictions_df[f"{target_col}_pred_GPR_LOOCV"] = y_pred_loocv
    predictions_df[f"{target_col}_error"] = (
        predictions_df[target_col] - predictions_df[f"{target_col}_pred_GPR_LOOCV"]
    )
    predictions_df[f"{target_col}_error_%"] = (
        predictions_df[f"{target_col}_error"] / predictions_df[target_col]
    ) * 100

    display(predictions_df)

    # Grid search results
    grid_results_df = pd.DataFrame(grid.cv_results_)

    grid_results_clean = grid_results_df[
        [
            "rank_test_score",
            "mean_test_score",
            "std_test_score",
            "param_regressor__gpr__kernel"
        ]
    ].copy()

    grid_results_clean = grid_results_clean.sort_values("rank_test_score")
    grid_results_clean["RMSE_LOOCV"] = -grid_results_clean["mean_test_score"]

    display(grid_results_clean)

    # Plot data
    plot_df = pd.DataFrame({
        "Sample_ID": data["Sample_ID"],
        f"Actual_{target_col}": y,
        f"Predicted_{target_col}": y_pred_loocv,
        f"Error_{target_col}": predictions_df[f"{target_col}_error"],
        f"Error_%_{target_col}": predictions_df[f"{target_col}_error_%"]
    })

    min_val = min(plot_df[f"Actual_{target_col}"].min(), plot_df[f"Predicted_{target_col}"].min())
    max_val = max(plot_df[f"Actual_{target_col}"].max(), plot_df[f"Predicted_{target_col}"].max())

    parity_df = pd.DataFrame({
        "Parity_x": [min_val, max_val],
        "Parity_y": [min_val, max_val]
    })

    # Actual vs predicted plot
    plt.figure(figsize=(6, 6))
    plt.scatter(plot_df[f"Actual_{target_col}"], plot_df[f"Predicted_{target_col}"])
    plt.plot([min_val, max_val], [min_val, max_val])
    plt.xlabel(f"Actual {target_label}")
    plt.ylabel(f"Predicted {target_label}")
    plt.title(f"GPR LOOCV - {target_label}")
    plt.grid(True)
    plt.tight_layout()
    plot_name = f"GPR_actual_vs_pred_{target_col}.png"
    plt.savefig(plot_name, dpi=300)
    plt.show()

    # Residual plot
    plt.figure(figsize=(6, 4))
    plt.scatter(
        plot_df[f"Predicted_{target_col}"],
        plot_df[f"Error_{target_col}"]
    )
    plt.axhline(0)
    plt.xlabel(f"Predicted {target_label}")
    plt.ylabel(f"Residual {target_label}")
    plt.title(f"GPR Residuals - {target_label}")
    plt.grid(True)
    plt.tight_layout()
    residual_plot_name = f"GPR_residual_{target_col}.png"
    plt.savefig(residual_plot_name, dpi=300)
    plt.show()

    # Train final model on all data
    final_model = best_model
    final_model.fit(X, y)

    return {
        "target_col": target_col,
        "target_label": target_label,
        "best_model": final_model,
        "metrics_df": metrics_df,
        "predictions_df": predictions_df,
        "grid_results_clean": grid_results_clean,
        "plot_df": plot_df,
        "parity_df": parity_df,
        "plot_name": plot_name,
        "residual_plot_name": residual_plot_name
    }


# ------------------------------------------------------------
# 5. Run two GPR models inside one framework
# ------------------------------------------------------------

result_sigma_m = run_single_target_gpr(
    target_col="sigma_m_MPa",
    target_label="σm (MPa)"
)

result_sigma_c = run_single_target_gpr(
    target_col="sigma_c_mean_MPa",
    target_label="σc mean (MPa)"
)

# ------------------------------------------------------------
# 6. Combined metrics and predictions
# ------------------------------------------------------------

combined_metrics = pd.concat(
    [
        result_sigma_m["metrics_df"],
        result_sigma_c["metrics_df"]
    ],
    ignore_index=True
)

combined_predictions = data.copy()

combined_predictions["sigma_m_pred_GPR_LOOCV"] = result_sigma_m["predictions_df"]["sigma_m_MPa_pred_GPR_LOOCV"]
combined_predictions["sigma_m_error_GPR"] = result_sigma_m["predictions_df"]["sigma_m_MPa_error"]
combined_predictions["sigma_m_error_%_GPR"] = result_sigma_m["predictions_df"]["sigma_m_MPa_error_%"]

combined_predictions["sigma_c_pred_GPR_LOOCV"] = result_sigma_c["predictions_df"]["sigma_c_mean_MPa_pred_GPR_LOOCV"]
combined_predictions["sigma_c_error_GPR"] = result_sigma_c["predictions_df"]["sigma_c_mean_MPa_error"]
combined_predictions["sigma_c_error_%_GPR"] = result_sigma_c["predictions_df"]["sigma_c_mean_MPa_error_%"]

print("\nCombined GPR metrics:")
display(combined_metrics)

print("\nCombined GPR predictions:")
display(combined_predictions)

# ------------------------------------------------------------
# 7. Save final framework
# ------------------------------------------------------------

gpr_framework = {
    "GPR_sigma_m": result_sigma_m["best_model"],
    "GPR_sigma_c": result_sigma_c["best_model"],
    "X_cols": X_cols,
    "y_cols": y_cols
}

joblib.dump(gpr_framework, "GPR_framework_final.pkl")

# ------------------------------------------------------------
# 8. Example prediction using the unified framework
# ------------------------------------------------------------

new_sample = pd.DataFrame({
    "Sludge_%": [20],
    "Plasticizer_%": [0.5],
    "Density": [0.00209]
})

pred_sigma_m = gpr_framework["GPR_sigma_m"].predict(new_sample[X_cols].values)
pred_sigma_c = gpr_framework["GPR_sigma_c"].predict(new_sample[X_cols].values)

print("\nPrediction for new sample using GPR framework:")
print(f"Predicted σm = {pred_sigma_m[0]:.3f} MPa")
print(f"Predicted σc = {pred_sigma_c[0]:.3f} MPa")

# ------------------------------------------------------------
# 9. Export all outputs to Excel
# ------------------------------------------------------------

with pd.ExcelWriter("GPR_publication_results.xlsx", engine="openpyxl") as writer:
    data.to_excel(writer, sheet_name="Clean_Data", index=False)

    combined_metrics.to_excel(writer, sheet_name="LOOCV_Metrics", index=False)
    combined_predictions.to_excel(writer, sheet_name="LOOCV_Predictions", index=False)

    result_sigma_m["grid_results_clean"].to_excel(writer, sheet_name="Grid_sigma_m", index=False)
    result_sigma_c["grid_results_clean"].to_excel(writer, sheet_name="Grid_sigma_c", index=False)

    result_sigma_m["plot_df"].to_excel(writer, sheet_name="Plot_Data_sigma_m", index=False)
    result_sigma_m["parity_df"].to_excel(writer, sheet_name="Parity_sigma_m", index=False)

    result_sigma_c["plot_df"].to_excel(writer, sheet_name="Plot_Data_sigma_c", index=False)
    result_sigma_c["parity_df"].to_excel(writer, sheet_name="Parity_sigma_c", index=False)

print("\nFiles saved:")
print("1. GPR_publication_results.xlsx")
print("2. GPR_actual_vs_pred_sigma_m_MPa.png")
print("3. GPR_actual_vs_pred_sigma_c_mean_MPa.png")
print("4. GPR_residual_sigma_m_MPa.png")
print("5. GPR_residual_sigma_c_mean_MPa.png")
print("6. GPR_framework_final.pkl")
```
